## Example Usage:

1. Run the client, send an individual message, and exit after receiving one response:

```python
!python selfdev_client.py --host "{XMPP_HOST}" --user "{XMPP_USER}" --password "{XMPP_PASSWORD}" --recipient "{XMPP_RECIPIENT}" --message "Hello!" --max-messages 1 --timeout 30 --allow-insecure
```

2. Join a MUC room, send a message to the room, for maximum messages or exits after timeout:

```python
! python selfdev_client.py --host "{XMPP_HOST}" --user "{XMPP_USER}" --password "{XMPP_PASSWORD}" --muc-host "{XMPP_MUC_HOST}" --join-rooms "{XMPP_JOIN_ROOMS}" --nick "{XMPP_NICK}" --message "echo Hello Room! @sh" --max-messages 30 --timeout 10 --allow-insecure # 2>/dev/null
```

## Example for Jupyter Notebooks:

You can also use this script programmatically in a Jupyter notebook:

```python
from selfdev_client import SelfdevClient
import asyncio
import nest_asyncio

nest_asyncio.apply()  # Allow nested event loops in Jupyter

XMPP_HOST="selfdev-prosody.dev.local"
XMPP_USER="selfdev_cli"
XMPP_PASSWORD="SDC-1254-Secret"
XMPP_RECIPIENT=f"ArtemArakcheev@{XMPP_HOST}"

client = SelfdevClient(
  jid=f"{XMPP_USER}@{XMPP_HOST}",
  password=XMPP_PASSWORD,
  recipient=XMPP_RECIPIENT,
  message="Hello privately from Jupyter!",
  max_messages=1,
  timeout=30,
  allow_insecure=True,
  verbose=True,
  output_format="{body}\n"
)

text = await asyncio.create_task(client.run())
print('\ntext:')
print(text)
```

```python
from selfdev_client import SelfdevClient
import asyncio
import nest_asyncio

nest_asyncio.apply()  # Allow nested event loops in Jupyter

XMPP_HOST="selfdev-prosody.dev.local"
XMPP_USER="selfdev_cli"
XMPP_PASSWORD="SDC-1254-Secret"
XMPP_MUC_HOST="conference.selfdev-prosody.dev.local"
XMPP_NICK=XMPP_USER

client = SelfdevClient(
  jid=f"{XMPP_USER}@{XMPP_HOST}",
  password=XMPP_PASSWORD,
  rooms=["command"],
  nick=XMPP_NICK,
  muc_host=XMPP_MUC_HOST,
  message="Hey @ls",
  max_messages=100,
  timeout=20,
  allow_insecure=True,
  verbose=True,
  output_format="[{nick}]: {body}\n"
)

text = await asyncio.create_task(client.run())
print('\ntext:')
print(text)
```
